```python
data = pd.read_csv('/content/train.csv')
print(data.shape)
data.head()
```

```python
print(data.duplicated().sum)

data.isnull().sum()
```

```python
data[['Context', 'Response']] = data[['Context', 'Response']].apply(lambda x: x.str.lower())
data.head()
```

```python
import re
def remove_tags(string):
    result = re.compile('<.*?>')
    if not isinstance(string, str):
        return string
    return re.sub(result, '', string)
data[['Context', 'Response']] = data[['Context', 'Response']].apply(lambda x: x.apply(remove_tags))
data.head()
```

```python
from types import new_class
import string
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
def remove_stopwords(string):
  if not isinstance(string, str):
    return string
  new_class = []
  for word in string.split():
    if word in stopwords.words('english'):
      new_class.append('')
    else:
      new_class.append(word)
  return " ".join(new_class)
data[['Context','Response']] = data[['Context','Response']].apply(lambda x: x.apply(remove_stopwords))
data.head()
```

```python
string.punctuation
def remove_punc(text):
  if not isinstance(text, str):
    return text
  return text.translate(str.maketrans('', '', string.punctuation))
data[['Context','Response']] = data[['Context','Response']].apply(lambda x: x.apply(remove_punc))
data.head()
```

```python
slang_word = {}
with open("slang.txt", 'r') as file:
  for line in file:
    line = line.strip()
    if not line:
      continue
    key,value = line.split('=',1)
    slang_word[key] = value
print(slang_word)
```

```python
def chat_conv(text):
  if not isinstance(text, str):
    return text
  new_text = []
  for w in text.split():
    if w.upper() in slang_word:
      new_text.append(slang_word[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
data[['Context','Response']] = data[['Context','Response']].apply(lambda x: x.apply(chat_conv))
data.head()
```

```python
from nltk.stem import WordNetLemmatizer
Le = WordNetLemmatizer()
nltk.download('wordnet')
def apply_le(text):
  if not isinstance(text, str):
      return text
  for w in text.split():
    return Le.lemmatize(text)
data[['Context','Response']] = data[['Context','Response']].apply(lambda x: x.apply(apply_le))
data.head()
```

```python
vocab =sorted({word for col in ['Context','Response']
               for text in data[col].dropna()
               for word in str(text).split()})
```

```python
word_to_index = {word:i for i,word in enumerate(vocab)}
print(word_to_index)
```

```python
print(len(vocab))
```

```python
def ohe(sentence):
  vectors = []
  for word in str(sentence).split():
    v = np.zeros(len(vocab))
    v[word_to_index[word]] = 1
    vectors.append(v)
  return np.array(vectors)
result = ohe(data['Context'][0])
print(result)
result.shape
```

```python
result = data[['Context', 'Response']].head(3).apply(lambda x: x.apply(ohe))
print(result['Context'].iloc[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=1000)
```

```python
from scipy.sparse import hstack

cv1 = CountVectorizer(max_features=1000)
cv2 = CountVectorizer(max_features=1000)

bow_context = cv1.fit_transform(data['Context'].fillna(''))
bow_response = cv2.fit_transform(data['Response'].fillna(''))

bow = hstack([bow_context, bow_response])
print(bow_response[0].toarray())
bow.shape
```

```python
words = (['context_' + w for w in cv1.get_feature_names_out()] +
         ['response_' + w for w in cv2.get_feature_names_out()])
counts = np.asarray(bow.sum(axis=0)).ravel()
word_counts = pd.DataFrame({'word': words, 'count': counts})
word_counts.sort_values(by='count', ascending=False)
word_counts.head(10)
```

```python
cv = CountVectorizer(ngram_range=(2,2))
bi_gram_context = cv.fit_transform(data['Context'].fillna(''))
bi_gram_response = cv.fit_transform(data['Response'].fillna(''))
bi_gram = hstack([bi_gram_context, bi_gram_response])
bi_gram.shape
```

```python
cv = CountVectorizer(ngram_range=(3,3))
tri_gram_context = cv.fit_transform(data['Context'].fillna(''))
tri_gram_response = cv.fit_transform(data['Response'].fillna(''))
tri_gram = hstack([tri_gram_context, tri_gram_response])
print(tri_gram.shape)
tri_gram[0].shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf1 = TfidfVectorizer(max_features=1000)
tf2 = TfidfVectorizer(max_features=1000)

tfidf_context = tf1.fit_transform(data['Context'].fillna(''))
tfidf_response = tf2.fit_transform(data['Response'].fillna(''))
tfidf = hstack([tfidf_context, tfidf_response])

words = (['context_' + w for w in tf1.get_feature_names_out()] +
         ['response_' + w for w in tf2.get_feature_names_out()])
idf_scores = np.concatenate([tf1.idf_, tf2.idf_])

word_idf = pd.DataFrame({'word': words, 'idf': idf_scores})
word_idf = word_idf.sort_values(by='idf', ascending=False)
print(word_idf.head(10))
```
